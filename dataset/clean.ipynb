### Image Extraction

This dataset includes mix of both normal and weaponized data. So we use exclude and include parameters to control and handle it.

```python
# input path of weaponized datasets
weaponized_input_image_directories = [
    "./raw_data/weaponized-image/train/",
    "./raw_data/weaponized-image/val",
]
```

```python
for path in weaponized_input_image_directories:
    # use includes to only save files that has "normal" inside filename
    extract_image(input_directory=path, output_directory="./cleaned-dataset/normal", includes="normal")
    # use excludes to save files where filename doesnot include "normal"
    extract_image(input_directory=path, output_directory="./cleaned-dataset/high-alert", excludes="normal")
```
